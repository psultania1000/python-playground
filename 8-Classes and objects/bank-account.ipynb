```python
import random
class BankAccount:
    def __init__(self, account_holder, initial_balance):
        self.account_holder = account_holder
        self.initial_balance = initial_balance
        self.account_number = random.randint(1, 10)
    
    def get_account_number(self):
        print(self.account_number)
        
    def deposit(self, amount):
        self.initial_balance += amount
        
    def withdrawal(self, amount):
        if amount > self.initial_balance:
            print("Transaction not valid")
        else:
            self.initial_balance -= amount
            
    def transfer(self, bank_account, amount):
        if amount > self.initial_balance:
            print("Transaction not valid")
        else:
            self.initial_balance -= amount
            bank_account.deposit(amount)
            
    def get_balance(self):
        print(self.initial_balance)
    
ankit = BankAccount("Ankit", 10000)
priyansh = BankAccount("Priyansh", 10000)
priyansh.transfer(ankit, 1000)

priyansh.get_balance()
ankit.get_balance()

```
