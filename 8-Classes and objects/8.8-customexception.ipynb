```python
class Error(Exception):
    pass

class dobException(Error):
    pass
```

```python
year = int(input("Enter the D.O.B."))

age = 2024-year

try:
    if(age<=30 and age>=20):
        print("The age is valid for exam")
    else:
        raise dobException
except dobException:
    print("Sorry, you are not eligible for the exam")
```
