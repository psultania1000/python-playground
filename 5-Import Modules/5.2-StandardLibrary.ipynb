### Standard Library Overview

Python's Standard Library is a vast collection of modules and packages that comes bundled with python, providing a wide range of functionalities out of the box. Here's an overview of some of the most commonly used modules and packages in the Python Standard Library.


```python
import array
arr = array.array('i', [1, 2, 3, 4])
print(arr)
```

```python
import math
print(math.pi)
```

```python
## random

import random
print(random.randint(1, 10))
```

```python
print(random.choice(['apple', 'banana', 'cherry']))
```

```python
## High level operations on files and collection

import shutil
shutil.copyfile('source.txt', 'destination.txt')
```

```python
## Data Serialization

import json
data={'name': 'Priyansh', 'age': 25}
json_str = json.dumps(data)

print(json_str)
print(type(json_str))
parsed_data = json.loads(json_str)
print(parsed_data)
print(type(parsed_data))
```

```python
## CSV

import csv

with open('example.csv', mode='w', newline='') as file:
    writer = csv.writer(file)
    writer.writerow(['name', 'age'])
    writer.writerow(['Krish', 32])
```

```python
with open('example.csv', mode='r') as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

```python
## datetime
from datetime import datetime, timedelta

now = datetime.now()
print(now)
```

```python
yesterday = now - timedelta(days=1)
print(yesterday)
```

```python
## time
import time
print(time.time())
time.sleep(2)
print(time.time())
```

```python
## Regular Expression
import re

pattern = r'\d+'
text = 'This is 13th page from 120'
match = re.search(pattern, text)
print(match.group())        # Returns first instance
```
