### Pandas - Dataframe and Series

Pandas is a powerful data manipulation library in Python, widely used for data analysis cleaning. It provides two primary data structure: Series and DataFrame. A Series is a one-dimensional array-like object, while a dataframe is a two dimensional, size-mutable and potentially heterogeneous tabular data structure with labeled axes (rows and columns)


```python
import sys
print(sys.executable)
```

```python
import pandas as pd

data = [1,2,3,4,5]
series = pd.Series(data)
print(series)
```

```python
## Create a series from dictionary elements

data={'a':1, 'b':2, 'c': 3}
series = pd.Series(data)
print(series)
```

```python
data = [10,20,30]
index=['a','b','c']

pd.Series(data, index=index)
```

```python
## Dataframe

## Create a dataframe from dict of list

data = {
    'Name': ['Priyansh', 'Utsav', 'Ashish'],
    'Age': [27, 26, 27],
    'City': ['Bangalore', 'Kolkata', 'Santragachi']
}

df = pd.DataFrame(data)
print(df)
```

```python
## Create a dataframe from a list of dictionary

data = [
    { 'Name': 'Priyansh', 'Age': 27, 'City': 'Bangalore'},
    { 'Name': 'Utsav', 'Age': 26, 'City': 'Kolkata'},
    { 'Name': 'Ashish', 'Age': 27, 'City': 'Santragachi'}
]

print(pd.DataFrame(data))
```
