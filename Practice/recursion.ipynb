```python
def printNum(n):
    if n == 0:
        return 0
    print(n)
    printNum(n-1)
```

```python
printNum(5)
```

```python
# factorial

def fact(n):
    if n == 1:
        return n
    return n * fact(n-1)
```

```python
print(fact(5))
```

```python
# Number of digits

def digits(n):
    if n < 10:
        return 1
    
    small_answer = digits(n//10)
    
    answer = 1 + small_answer
    
    return answer

print(digits(12334))
```
