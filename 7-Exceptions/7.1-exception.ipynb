### Understanding Exceptions

Exception handling in Python allows you to handle errors gracefully and take corrective actions without stopping the execution of the program. This lesson will cover the basics of exceptions, including how to use try, except, else and finally blocks

### What are Exceptions?

Exceptions are events that disrupts the normal flow of a program. They occur when an error is encountered during program execution. Common exceptions include:

- ZeroDivisionError: Dividing by Zero
- FileNotFoundError: File not found
- ValueError: Invalid value
- TypeError: Invalid Error


```python
## Exception using try and except

try: 
    a=b
except:
    print("variable has not been assigned")
```

```python
## Using same exception class
## Other Exception wont be caught if we specify a specific class for it use 'Exception'
## try except only throws first exception, if there are multiple ones.
try: 
    a=b
except NameError as ex:
    print(ex)
```

```python
try:
    # result = 1/0
    a = b
except ZeroDivisionError as ex:
    print(ex)
except Exception as ex1:
    print(ex1)    
```

```python
try:
    num=int(input("enter a number"))
    result=10/num
except ValueError as ex:
    print(ex)
except ZeroDivisionError as ex1:
    print(ex1)
except Exception as ex2:
    print(ex2)
else:
    print(result)
```

```python
## try, except, else and finally

try:
    num=int(input("enter a number"))
    result=10/num
except ValueError as ex:
    print(ex)
except ZeroDivisionError as ex1:
    print(ex1)
except Exception as ex2:
    print(ex2)
else:
    print(result)
finally:
    print("Execution complete")
```

```python
### File handling and exception handling

try:
    file=open('example1.txt', 'r')
    content=file.read()
    print(content)
except FileNotFoundError as fe:
    print(fe)
finally:
    if 'file' in locals() or not file.closed():
        file.close()
        print('file close')
    
```
