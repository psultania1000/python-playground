### Decorators

Decorators are a powerful and flexible feature in python that allows you to modify the behavior of a function of a function or class method. They are commonly used to add functionality to functions or methods without modifying their actual code. This lesson covers the basics of decorators, including how to create and use them.

```python
## Closures

def main_welcome(msg):
    # msg="Welcome"
    
    def sub_welcome_method():
        print("Welcome to advance python course")
        print(msg)
        print("Please learn this concepts properly")
        
    return sub_welcome_method()
```

```python
## passing function

def main_welcome(func, arg):
    # msg="Welcome"
    
    def sub_welcome_method():
        print("Welcome to advance python course")
        print(func(arg))
        print("Please learn this concepts properly")
        
    return sub_welcome_method()
```

```python
main_welcome(print, "Welcome priyansh")
```

```python
main_welcome(len, [1,2,3,4,5,6])
```

```python
### Decorators

def main_welcome(func):
    # msg="Welcome"
    
    def sub_welcome_method():
        print("Welcome to advance python course")
        func()
        print("Please learn this concepts properly")
        
    return sub_welcome_method()
```

```python

def course_introduction():
    print("This is an advance python course")
    
course_introduction()
```

```python
main_welcome(course_introduction)
```

```python
@main_welcome  # decorator
def course_introduction():
    print("This is an advance python course")
```

```python
## Decorator

def my_decorator(func):
    def wrapper():
        print("Something is happening before the function is called.")
        func()
        print("Something is happening after the function is called")
    return wrapper
```

```python
@my_decorator
def say_hello():
    print("Hello!")
```

```python
say_hello()
```

```python
## Decorators with arguments

def repeat(n):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for _ in range(n):
                func(*args, **kwargs)
        return wrapper
    return decorator
```

```python
 @repeat(3)
 def say_hello():
     print("Hello")
```

```python
say_hello()
```
